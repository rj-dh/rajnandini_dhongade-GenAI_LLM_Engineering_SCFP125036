topic -> story -> poem -> joke

```python
from langchain_groq import ChatGroq
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=os.environ['API_KEY']
)
```

```python
llm.invoke("hi")
```

```python
class State(TypedDict):
    topic : str
    story : str
    poem : str
    joke : str
```

```python
def story(state : State):
    #print(state['topic'])
    response=llm.invoke(f"write a stroy on this {state["topic"]}")
    return {"story":response.content}
```

```python
d = {"topic":"AI"}
```

```python
topic_res=story(d)
```

```python
#story = d
```

```python
def poem(state: State):
    print(state['story'])

    response = llm.invoke(
        f"Write a 5-line poem based on this story: {state['story']}"
    )

    return {"poem": response.content}

poem_res = poem(topic_res)
```

```python
def joke(state : State):
    # print(sate['Story'])
    response = llm.invoke(f"Write a short joke based on this poem: {state['poem']}")   
    return {"joke":response.content}
```

```python
joke(poem_res)
```

```python
from langgraph.graph import StateGraph, START, END
```

```python
print("story:", type(story))
print("poem:", type(poem))
print("joke:", type(joke))
```

```python
workflow=StateGraph(State)
```

```python
workflow.add_node("story",story)
workflow.add_node("poem", poem)
workflow.add_node("joke", joke)
```

```python
workflow.add_edge(START,"story")
workflow.add_edge("story","poem")
workflow.add_edge("poem","joke")
workflow.add_edge("joke", END)
```

```python
app = workflow.compile()
```
