```python
from langchain_core.tools import tool
```

```python
@tool
def convert_marks_to_cgpa(marks:float) -> float:
    """this function convert marks into cgpa""" #docstring
    return round(marks/9.6)
```

```python
@tool
def addition(num1:int,num2:int) -> int:
    """this function adds 2 numbers"""
    return (num1 + num2)
```

```python
addition.invoke({"num1":4,"num2":6})
```

```python
from langchain_core.messages import AIMessage,HumanMessage,ToolMessage
```

```python
import builtins
```

```python
message = []

while True:
    user_query = builtins.input("Enter user query: ")

    if user_query.lower() == "exit":
        break

    message.append(HumanMessage(content=user_query))

    ai_response = llm_with_tool.invoke(message)

    if ai_response.tool_calls:
        message.append(ai_response)

        for tool_call in ai_response.tool_calls:
            tool_args = tool_call["args"]

            if tool_call["name"] == add.name:
                tool_response = add.invoke(tool_args)
            else:
                raise ValueError(f"Unknown tool: {tool_call['name']}")

            message.append(
                ToolMessage(
                    content=str(tool_response),
                    tool_call_id=tool_call["id"]
                )
            )

        response = llm_with_tool.invoke(message)
        message.append(response)
        print(response.content)

    else:
        message.append(ai_response)
        print(ai_response.content)
```
