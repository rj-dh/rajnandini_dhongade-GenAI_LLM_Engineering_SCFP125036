```python
from typing import TypedDict
```

```python
import os
```

```python
from langchain_groq import ChatGroq
```

```python
class AssignmentState(TypedDict):
    Assignment_topic: str
    assignment_text: str
    assignment_summary: str
    assignment_evaluation: str
    assignment_feedback: str
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=os.environ["GROQ_API_KEY"]
)
```

```python
def assignment_summary(state: AssignmentState):

    response = llm.invoke(
        f"""
        Based on the assignment topic:
        {state['Assignment_topic']}

        Summarize the following assignment:
        {state['assignment_text']}
        """
    )

    return {"assignment_summary": response.content}
```

```python
def assignment_evaluation(state: AssignmentState):

    response = llm.invoke(f"""Evaluate the assignment based on:Assignment Topic:{state['Assignment_topic']}
            Assignment Summary:{state['assignment_summary']}Marking Criteria:Check relevance, accuracy, completeness,clarity and organization.
            Provide a detailed evaluation.""")

    return {"assignment_evaluation": response.content}
```

```python
def assignment_feedback(state: AssignmentState):

    response = llm.invoke(f"""Based on the assignment evaluation:{state['assignment_evaluation']}
            Prepare constructive""" )

    return {"assignment_feedback": response.content}
```

```python
from langgraph.graph import StateGraph,START,END
```

```python
workflow = StateGraph(AssignmentState)

workflow.add_node("assignment_summary", assignment_summary)
workflow.add_node("assignment_evaluation", assignment_evaluation)
workflow.add_node("assignment_feedback", assignment_feedback)

workflow.add_edge(START, "assignment_summary")
workflow.add_edge("assignment_summary", "assignment_evaluation")
workflow.add_edge("assignment_evaluation", "assignment_feedback")
workflow.add_edge("assignment_feedback", END)

app = workflow.compile()
```

```python
result = app.invoke({
    "Assignment_topic": "Artificial Intelligence",
    "assignment_text": """
    Artificial Intelligence is a technology that enables
    machines to perform tasks that normally require
    human intelligence. It is used in healthcare,
    education and transportation.
    """
})

print("SUMMARY:")
print(result["assignment_summary"])

print("\nEVALUATION:")
print(result["assignment_evaluation"])

print("\nFEEDBACK:")
print(result["assignment_feedback"])
```
