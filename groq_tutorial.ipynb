```python
def ask_llm(user_query):
    chat_history.append({"role":"user",
                    "content":user_query})
    
    message = chat_history

    chat_completion = client.chat.completions.create(messages = message, model="openai/gpt-oss-120b")
    response = chat_completion.choices[0].message.content

    chat_history.append({"role":"assistant",
                    "content":response})
    return response
```

```python
def ask_llm_temp_max(user_query, max_token = 300):
    message = [
        {
            "role":"user",
            "content":user_query
    }    
]
    chat_completion = client.chat.completions.create(messages = message, model="openai/gpt-oss-120b", temperature = 1, max_tokens = max_token)
    response = chat_completion.choices[0].message.content
    return response
```
